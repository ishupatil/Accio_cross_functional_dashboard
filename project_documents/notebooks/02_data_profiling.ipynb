### 2. Load Raw CSV Files
Placeholder for load raw csv files steps.

```python
import glob
import os

# Find all raw CSV files we extracted
raw_files = glob.glob("../data/raw/*.csv")
print(f"Found {len(raw_files)} raw files to profile:\n")

# Load into a dictionary: key is the clean filename, value is the DataFrame
dfs = {}
for file in raw_files:
    name = os.path.splitext(os.path.basename(file))[0]
    dfs[name] = pd.read_csv(file)
    print(f" - Loaded: {name} (Shape: {dfs[name].shape})")
```

### 3. Dataset Shapes & Info
Placeholder for dataset shapes & info steps.

```python
# Create a summary table showing row counts, column counts, and memory usage
shape_summary = []
for name, df in dfs.items():
    shape_summary.append({
        "Dataset Name": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Memory Usage (KB)": round(df.memory_usage(deep=True).sum() / 1024, 2)
    })

df_summary = pd.DataFrame(shape_summary).sort_values(by="Rows", ascending=False)
df_summary
```

### 4. Missing Value Analysis
Placeholder for missing value analysis steps.

```python
# Identify columns containing null/missing values across all datasets
missing_summary = []
for name, df in dfs.items():
    null_counts = df.isnull().sum()
    null_cols = null_counts[null_counts > 0]
    
    for col, count in null_cols.items():
        missing_summary.append({
            "Dataset": name,
            "Column": col,
            "Missing Rows": count,
            "Percentage (%)": round((count / len(df)) * 100, 2)
        })
        
df_missing = pd.DataFrame(missing_summary)
if df_missing.empty:
    print("No missing values found in any dataset!")
else:
    print(f"Found {len(missing_summary)} columns with missing values across datasets:\n")
    display(df_missing.sort_values(by="Percentage (%)", ascending=False))
```

### 5. Check Data Types and Anomalies
Placeholder for check data types and anomalies steps.

```python
# Profile the main transactions table for data types, distributions, and possible anomalies
orders_df = dfs.get("sales_transaction_orders")
if orders_df is not None:
    print("--- Profiling Core Table: sales_transaction_orders ---\n")
    
    print("1. Data Types and Sample:")
    display(orders_df.info())
    
    print("\n2. Summary Statistics for Numeric Fields:")
    display(orders_df.describe())
    
    print("\n3. Checking for Negative Numeric Anomalies:")
    neg_net = orders_df[orders_df["net_amount"] < 0]
    neg_total = orders_df[orders_df["total_amount"] < 0]
    print(f" - Orders with negative net_amount: {len(neg_net)}")
    print(f" - Orders with negative total_amount: {len(neg_total)}")
    
    print("\n4. Check Unique Counts & Distribution of Payment Modes:")
    print(orders_df["payment_mode"].value_counts(dropna=False))
else:
    print("Core orders dataset not found. Make sure sales_transaction_orders.csv exists.")
```
